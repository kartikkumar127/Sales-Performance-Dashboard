## 1. Project Objective

The objective of this project is to analyze retail sales data and identify key trends and business insights related to sales, profit, customers, products, categories, and regions.

The dataset used in this project is the Superstore sales dataset sourced from Kaggle.

The dataset contains retail transaction information such as orders, customers, products, sales, quantity, discount, profit, category, sub-category, and region.

## 7. Data Cleaning

### Standardizing Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

df.columns
```

### cleaning text columns

```python
text_columns = df.select_dtypes(include=["str", "object"]).columns

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
```

### Converting Date Columns

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["ship_date"] = pd.to_datetime(df["ship_date"], errors="coerce")
```

```python
df[["order_date", "ship_date"]].dtypes
```

```python
df.isnull().sum()
```

### Removing Duplicate Records

```python
df = df.drop_duplicates()

df.duplicated().sum()
```

### Validating Numerical Columns

```python
numeric_columns = [
    "sales",
    "quantity",
    "discount",
    "profit"
]

df[numeric_columns].describe()
```

## 8. Feature Engineering

### Calculating Shipping Days

```python
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days

df[["order_date", "ship_date", "shipping_days"]].head()
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["month"] = df["order_date"].dt.month
```

```python
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)
```

```python
df["profit_margin"] = (
    df["profit"] / df["sales"]
) * 100
```

```python
df["profit_margin"] = df["profit_margin"].replace(
    [np.inf, -np.inf],
    np.nan
)
```

### Sales and Profit by Category

```python
category_analysis = (
    df.groupby("category")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum"),
        quantity=("quantity", "sum")
    )
    .sort_values("sales", ascending=False)
)

category_analysis
```

### Sales and Profit by Region

```python
region_analysis = (
    df.groupby("region")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum")
    )
    .sort_values("sales", ascending=False)
)

region_analysis
```

### Sub-Category Performance

```python
subcategory_analysis = (
    df.groupby("sub_category")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum")
    )
    .sort_values("sales", ascending=False)
)

subcategory_analysis
```

### Top 10 Products by Sales

```python
top_products = (
    df.groupby("product_name")["sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

### Top 10 Products by profit

```python
top_profit_products = (
    df.groupby("product_name")["profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_profit_products
```

### Loss-Making Products

```python
loss_products = (
    df.groupby("product_name")["profit"]
    .sum()
    .sort_values()
)

loss_products.head(10)
```

### Monthly Sales Trend

```python
monthly_sales = (
    df.groupby("year_month")["sales"]
    .sum()
    .reset_index()
)

monthly_sales.head()
```

### Monthly Profit

```python
monthly_profit = (
    df.groupby("year_month")["profit"]
    .sum()
    .reset_index()
)

monthly_profit.head()
```

### Customer Segment Performance

```python
segment_analysis = (
    df.groupby("segment")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("sales", ascending=False)
)

segment_analysis
```

### Discount and Profit Analysis

```python
discount_analysis = (
    df.groupby("discount")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum")
    )
    .sort_index()
)

discount_analysis
```

### Shipping Performance

```python
shipping_analysis = (
    df.groupby("ship_mode")
    .agg(
        orders=("order_id", "nunique"),
        average_shipping_days=("shipping_days", "mean")
    )
)

shipping_analysis
```

# 10. Business Insights

### 1. Sales Performance by Category
- Technology generated the highest sales at $836,154.03 and the highest profit at

   $145,454.95.

  
- Office Supplies had the highest quantity sold with 22,906 units and generated

  $122,490.80 in profit.

  
- Furniture generated $741,999.80 in sales but only

  $18,451.27 in profit, indicating comparatively lower profitability.

### 2. Regional Profitability
- The West region generated the highest sales at $725,457.82 and the highest profit at

  $108,418.45.

- The East region ranked second in sales with $678,781.24 and generated

  $91,522.78 in profit.
  
- The Central region generated $501,239.89 in sales and

  $39,706.36 in profit.

  
- The South region had the lowest sales at $391,721.91, but generated

  $46,749.43 in profit, which is higher than Central's profit despite having lower sales.

### 3. Customer Segment Performance
- Consumer is the largest customer segment, contributing approximately $1.16M in sales.
  
- Corporate contributes approximately $0.71M, while Home Office contributes approximately

  $0.43M.
  
- Consumer customers therefore represent the largest revenue opportunity.

### 4. Product Performance
- The Canon imageCLASS 2200 Advanced Copier is the top product by sales and also the top product by profit in the analysis.
- A small number of products contribute substantially more revenue than the remaining products, highlighting the importance of identifying and maintaining high-performing products.
- Loss-making products were identified through product-level profitability analysis and should be reviewed for pricing, discounting, and cost-related issues.

### 5. Discount and Profitability
- The discount-versus-profitability analysis shows that higher discount levels are generally associated with weaker profitability.
- Several high-discount observations have low or negative profit.
- Discount policies should therefore be monitored to ensure that increased sales volume does not reduce overall profitability.

### 6. Shipping Performance
- Standard Class accounts for the majority of orders, with approximately 2,994 orders.
- Standard Class also has the highest average shipping time at approximately 5 days.
- Same Day has the shortest shipping time, while First Class and Second Class provide intermediate delivery times.

### 7. Overall Performance
- The business generated approximately $2.30M in total sales and

  $286K in total profit.
  
- Overall profit margin is approximately 12.47%.
- The analysis indicates that profitability is influenced by product selection, regional performance, customer segments, discount levels, and shipping methods.

# 11.Business Recommendations

1. **Focus on high-performing categories**
   - Continue investing in Technology while monitoring the performance of Furniture and Office Supplies.

2. **Improve regional profitability**
   - Investigate the lower profitability of Central and South regions by analyzing product mix, pricing, discounts, and customer demand.

3. **Review loss-making products**
   - Evaluate pricing, discount rates, and costs for products generating negative profit.

4. **Optimize discount strategy**
   - Monitor high-discount transactions and establish discount thresholds that protect profit margins.

5. **Improve shipping efficiency**
   - Review Standard Class shipping performance and investigate opportunities to reduce average delivery time without significantly increasing logistics costs.

6. **Prioritize valuable customer segments**
   - Continue focusing on the Consumer segment while identifying opportunities to increase revenue from Corporate and Home Office customers.
